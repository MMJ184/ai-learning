````python
# Run this cell first (Shift+Enter). It finds your project folder and defines `check`,
# which prints "Correct" when an exercise works.
import math
from pathlib import Path

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists())
DATA = ROOT / "data"


def _same(a, b, tol=1e-6):
    import math as _math
    import numbers as _numbers
    try:
        import numpy as _np
    except ImportError:
        _np = None
    if _np is not None and (isinstance(a, _np.ndarray) or isinstance(b, _np.ndarray)):
        a, b = _np.asarray(a), _np.asarray(b)
        if a.shape != b.shape:
            return False
        if a.dtype.kind in "biuf" and b.dtype.kind in "biuf":
            return bool(_np.allclose(a, b, rtol=1e-5, atol=tol))
        return bool((a == b).all())
    if isinstance(a, bool) or isinstance(b, bool):
        return a == b
    if isinstance(a, _numbers.Number) and isinstance(b, _numbers.Number):
        return _math.isclose(a, b, rel_tol=1e-5, abs_tol=tol)
    if isinstance(a, (list, tuple)) and isinstance(b, (list, tuple)):
        return len(a) == len(b) and all(_same(x, y, tol) for x, y in zip(a, b))
    if isinstance(a, dict) and isinstance(b, dict):
        return a.keys() == b.keys() and all(_same(a[k], b[k], tol) for k in a)
    return a == b


def check(label, actual, expected=True, tol=1e-6):
    """Run `actual` (usually a lambda) and print Correct if it equals `expected`."""
    try:
        got = actual() if callable(actual) else actual
    except NotImplementedError:
        print(f"⏳ Not done yet: {label}")
        return
    except Exception as error:  # show the problem without stopping the notebook
        print(f"❌ {label}: your code raised {type(error).__name__}: {error}")
        return
    if _same(got, expected, tol):
        print(f"✅ Correct: {label}")
    else:
        print(f"❌ {label}\n   expected: {expected!r}\n   got:      {got!r}")

print("Setup works")
````

````python
# 2.1 FizzBuzz, returned as a list
def fizzbuzz(n):
    """Return a list of strings for 1..n: "Fizz" for multiples of 3, "Buzz" for multiples of 5,
    "FizzBuzz" for multiples of both, otherwise the number as a string."""
    result = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            result.append("FizzBuzz")
        elif i % 3 == 0:
            result.append("Fizz")
        elif i % 5 == 0:
            result.append("Buzz")
        else:
            result.append(str(i))
    return result

check("fizzbuzz(5)", lambda: fizzbuzz(5), ["1", "2", "Fizz", "4", "Buzz"])
check("15 is FizzBuzz", lambda: fizzbuzz(15)[-1], "FizzBuzz")
check("fizzbuzz(30) has 30 items", lambda: len(fizzbuzz(30)), 30)
````

````python
# 2.2 Temperature converter with f-strings
def c_to_f(celsius):
    """Convert Celsius to Fahrenheit: F = C × 9/5 + 32."""
    return celsius * 9 / 5 + 32


def describe_temperature(celsius):
    """Return text like '30°C is 86.0°F', with one decimal place for Fahrenheit."""
    return f"{celsius}°C is {c_to_f(celsius):.1f}°F"

check("100°C", lambda: c_to_f(100), 212)
check("-40°C", lambda: c_to_f(-40), -40)
check("describe_temperature(30)", lambda: describe_temperature(30), "30°C is 86.0°F")
````

````python
# 2.3 Count vowels
def count_vowels(text):
    """Count the vowels (a, e, i, o, u) in text, ignoring upper/lower case."""
    return sum(1 for letter in text.lower() if letter in "aeiou")

check("Machine Learning", lambda: count_vowels("Machine Learning"), 6)
check("PYTHON", lambda: count_vowels("PYTHON"), 1)
check("empty text", lambda: count_vowels(""), 0)
````

````python
# 2.4 A while loop
def sum_even_below(n):
    """Add up the even numbers from 0 up to (not including) n, using a while loop."""
    total, i = 0, 0
    while i < n:
        total += i
        i += 2
    return total

check("sum_even_below(10)", lambda: sum_even_below(10), 20)
check("sum_even_below(1)", lambda: sum_even_below(1), 0)
check("sum_even_below(101)", lambda: sum_even_below(101), 2550)
````

## Night 3 · Wed 25 Nov: Lists, dicts, sets and comprehensions
Slices stop *before* the end index, like `range`. A comprehension replaces `map` + `filter`.

````python
nums = list(range(10))  # [0, 1, 2, ..., 9]
print(nums)
````

````python
# 3.1 Slicing: one line each, no loops
def first_three(items):
    return items[:3]


def last_two(items):
    return items[-2:]


def every_other(items):
    """Items at positions 0, 2, 4, ..."""
    return items[::2]


def reversed_copy(items):
    return items[::-1]

check("first_three", lambda: first_three(nums), [0, 1, 2])
check("last_two", lambda: last_two(nums), [8, 9])
check("every_other", lambda: every_other(nums), [0, 2, 4, 6, 8])
check("reversed_copy", lambda: reversed_copy(nums), [9, 8, 7, 6, 5, 4, 3, 2, 1, 0])
````

````python
# 3.2 One list comprehension
def squares_of_odds(n):
    """Return the squares of the odd numbers from 1 to n: squares_of_odds(5) is [1, 9, 25]."""
    return [x * x for x in range(1, n + 1) if x % 2 == 1]

check("squares_of_odds(5)", lambda: squares_of_odds(5), [1, 9, 25])
check("squares_of_odds(0)", lambda: squares_of_odds(0), [])
````

````python
# 3.3 A dict as a counter
def word_counts(text):
    """Return a dict mapping each lowercase word to how many times it appears. Split on spaces."""
    counts = {}
    for word in text.lower().split():
        counts[word] = counts.get(word, 0) + 1
    return counts

check("word_counts", lambda: word_counts("The cat and the hat"), {"the": 2, "cat": 1, "and": 1, "hat": 1})
````

````python
# 3.4 Sets
posts = [["python", "ai"], ["ai", "ml"], ["python", "web"], ["ml"]]


def unique_tags(posts):
    """Return the sorted list of tags used in any post, each listed once."""
    return sorted({tag for post in posts for tag in post})


def shared_tags(a, b):
    """Return the sorted list of tags that appear in both lists a and b."""
    return sorted(set(a) & set(b))

check("unique_tags", lambda: unique_tags(posts), ["ai", "ml", "python", "web"])
check("shared_tags", lambda: shared_tags(["python", "ai"], ["ai", "ml"]), ["ai"])
````

````python
# 3.5 Sorting with a key
students = [
    {"name": "Asha", "score": 82},
    {"name": "Ravi", "score": 95},
    {"name": "Meera", "score": 78},
    {"name": "Kabir", "score": 95},
]


def ranking(students):
    """Return the names sorted by score, highest first. Equal scores go in alphabetical order."""
    ordered = sorted(students, key=lambda s: (-s["score"], s["name"]))
    return [s["name"] for s in ordered]

check("ranking", lambda: ranking(students), ["Kabir", "Ravi", "Asha", "Meera"])
````

````python
# 4.1 Default and keyword arguments
def make_greeting(name, greeting="Hello", punctuation="!"):
    """Return e.g. 'Hello, Asha!'. Callers can change greeting and punctuation by keyword."""
    return f"{greeting}, {name}{punctuation}"

check("defaults", lambda: make_greeting("Asha"), "Hello, Asha!")
check("punctuation by keyword", lambda: make_greeting("Ravi", punctuation="?"), "Hello, Ravi?")
check("greeting by keyword", lambda: make_greeting("Meera", greeting="Namaste"), "Namaste, Meera!")
````

````python
# 4.2 *args
def average(*numbers):
    """Return the mean of any number of arguments, or 0.0 when there are none."""
    return sum(numbers) / len(numbers) if numbers else 0.0

check("average(2, 4, 9)", lambda: average(2, 4, 9), 5.0)
check("average()", lambda: average(), 0.0)
check("spreading a list: average(*[1, 2])", lambda: average(*[1, 2]), 1.5)
````

````python
# 4.3 **kwargs
def build_profile(name, **details):
    """Return a dict with 'name' plus every keyword argument given."""
    return {"name": name, **details}

check(
    "build_profile",
    lambda: build_profile("Asha", city="Pune", language="Python"),
    {"name": "Asha", "city": "Pune", "language": "Python"},
)
````

````python
# 4.4 A lambda as a sort key
def by_length(words):
    """Return the words sorted from shortest to longest; equal lengths go alphabetically."""
    return sorted(words, key=lambda word: (len(word), word))

check("by_length", lambda: by_length(["banana", "kiwi", "apple", "fig"]), ["fig", "kiwi", "apple", "banana"])
````

````python
# 4.5 A closure: a function that builds functions
def make_multiplier(factor):
    """Return a new function that multiplies its input by factor."""
    def multiply(x):
        return x * factor
    return multiply

check("make_multiplier(2)(21)", lambda: make_multiplier(2)(21), 42)
check("make_multiplier(10)(3)", lambda: make_multiplier(10)(3), 30)
````

````python
# 4.6 Type hints: they document, Python doesn't enforce them
def describe(numbers: list[float]) -> dict[str, float]:
    """Return {'min': ..., 'max': ..., 'mean': ...} for a non-empty list of numbers."""
    return {"min": min(numbers), "max": max(numbers), "mean": sum(numbers) / len(numbers)}

check("describe", lambda: describe([3, 1, 4, 1, 5]), {"min": 1, "max": 5, "mean": 2.8})
````

## Night 5 · Fri 27 Nov: Classes, errors, files and modules
`self` is an explicit `this`, `__init__` is the constructor, and `with` closes files for you.

````python
# 5.1 A class that raises errors
class BankAccount:
    """An account with an owner and a balance that starts at 0.

    deposit(amount) adds money and withdraw(amount) removes it. Both raise ValueError
    for amounts of 0 or less; withdraw also raises ValueError when the balance is too low.
    """

    def __init__(self, owner):
        self.owner = owner
        self.balance = 0

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self.balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal must be positive")
        if amount > self.balance:
            raise ValueError("Insufficient funds")
        self.balance -= amount

def _deposit_then_withdraw():
    account = BankAccount("Asha")
    account.deposit(500)
    account.withdraw(200)
    return account.balance


def _overdraft_raises():
    account = BankAccount("Asha")
    account.deposit(100)
    try:
        account.withdraw(500)
    except ValueError:
        return True
    return False


check("deposit 500 then withdraw 200 leaves 300", _deposit_then_withdraw, 300)
check("withdrawing too much raises ValueError", _overdraft_raises, True)
````

````python
# 5.2 A dataclass writes __init__ and __eq__ for you
from dataclasses import dataclass


@dataclass
class Point:
    x: float
    y: float

    def distance_to(self, other):
        """Return the straight-line distance to another Point. Hint: math.hypot."""
        return math.hypot(self.x - other.x, self.y - other.y)

check("distance from (0, 0) to (3, 4)", lambda: Point(0, 0).distance_to(Point(3, 4)), 5.0)
check("dataclasses compare by value", lambda: Point(1, 2) == Point(1, 2), True)
````

````python
# 5.3 try / except instead of checking first
def parse_number(text):
    """Return text as a float, or None if it isn't a number."""
    try:
        return float(text)
    except ValueError:
        return None

check("'3.5'", lambda: parse_number("3.5"), 3.5)
check("' 42 '", lambda: parse_number(" 42 "), 42.0)
check("'abc'", lambda: parse_number("abc"), None)
````

````python
# 5.4 Files and JSON with `with open(...)`
import json


def save_progress(path, progress):
    """Write the dict `progress` to `path` as JSON."""
    with open(path, "w", encoding="utf-8") as file:
        json.dump(progress, file, indent=2)


def load_progress(path):
    """Read the JSON file at `path` and return the dict."""
    with open(path, encoding="utf-8") as file:
        return json.load(file)

def _round_trip():
    path = ROOT / "progress.json"
    save_progress(path, {"night": 5, "done": ["setup", "syntax"]})
    return load_progress(path)


check("save then load gives the same dict", _round_trip, {"night": 5, "done": ["setup", "syntax"]})
````

````python
# 6.1 collections.Counter
from collections import Counter


def top_words(text, n=3):
    """Return the n most common lowercase words as (word, count) pairs."""
    return Counter(text.lower().split()).most_common(n)

check("top_words", lambda: top_words("a b a c b a", 2), [("a", 3), ("b", 2)])
````

````python
# 6.2 pathlib
def csv_files(folder):
    """Return the sorted file names (not full paths) of the .csv files in folder. Hint: Path(folder).glob("*.csv")"""
    return sorted(path.name for path in Path(folder).glob("*.csv"))

check("the three datasets in data/", lambda: csv_files(DATA), ["penguins.csv", "tips.csv", "titanic.csv"])
````

````python
# 6.3 datetime
from datetime import date


def days_until(target, today):
    """Return how many days from `today` until `target` (both datetime.date)."""
    return (target - today).days

check("23 Nov to the Month 1 project on 20 Dec", lambda: days_until(date(2026, 12, 20), date(2026, 11, 23)), 27)
````

````python
# 6.4 A generator: values are produced one at a time with `yield`
def countdown(n):
    """Yield n, n - 1, ..., 1."""
    while n > 0:
        yield n
        n -= 1

check("list(countdown(3))", lambda: list(countdown(3)), [3, 2, 1])
````

## Night 7 · Sun 29 Nov: Review and project: a text analyzer
Finish the three functions until every check passes, then run your analyzer on a whole book.

````python
import json
import re
from collections import Counter

STOP_WORDS = {
    "a", "about", "after", "all", "am", "an", "and", "any", "are", "as", "at", "be", "been", "before",
    "being", "but", "by", "can", "could", "did", "do", "does", "for", "from", "had", "has", "have",
    "he", "her", "him", "his", "how", "i", "if", "in", "into", "is", "it", "its", "may", "me", "more",
    "much", "must", "my", "no", "not", "now", "of", "on", "one", "only", "or", "our", "out", "said",
    "she", "should", "so", "some", "such", "than", "that", "the", "their", "them", "then", "there",
    "they", "this", "to", "too", "up", "upon", "us", "very", "was", "we", "were", "what", "when",
    "which", "who", "will", "with", "would", "you", "your",
}
sample = "The cat sat. The cat ran! Did the dog run?"
````

````python
# 7.1 Clean words with a regular expression. Hint: re.findall(r"[a-z']+", text.lower())
def clean_words(text):
    """Return the lowercase words in text, keeping only letters and apostrophes.
    "Hello, World! It's 2026." -> ['hello', 'world', "it's"]"""
    return re.findall(r"[a-z']+", text.lower())


# 7.2 Count sentences
def sentence_count(text):
    """Count sentences: runs of . ! or ? each end one sentence ("Wait... what?!" is 2)."""
    return len(re.findall(r"[.!?]+", text))


# 7.3 Put it together
def analyze(text, top_n=5):
    """Return a dict with:
    'words': total words, 'unique_words': distinct words, 'sentences': sentence count,
    'avg_word_length': mean word length rounded to 2 decimals,
    'top_words': the top_n most common words NOT in STOP_WORDS, as [word, count] lists."""
    words = clean_words(text)
    content_words = [word for word in words if word not in STOP_WORDS]
    return {
        "words": len(words),
        "unique_words": len(set(words)),
        "sentences": sentence_count(text),
        "avg_word_length": round(sum(len(word) for word in words) / len(words), 2) if words else 0.0,
        "top_words": [[word, count] for word, count in Counter(content_words).most_common(top_n)],
    }

check("clean_words", lambda: clean_words("Hello, World! It's 2026."), ["hello", "world", "it's"])
check("sentence_count", lambda: sentence_count("Wait... what?! Yes."), 3)
check(
    "analyze(sample, top_n=2)",
    lambda: analyze(sample, top_n=2),
    {"words": 10, "unique_words": 7, "sentences": 3, "avg_word_length": 3.0, "top_words": [["cat", 2], ["sat", 1]]},
)
````
